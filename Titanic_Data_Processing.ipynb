```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the Titanic dataset
df = pd.read_csv("titanic.csv")
print(df.head())
```

```python
# Check dataset information and missing values
print(df.shape)
print(df.info())
print(df.isnull().sum())
print(df.describe())
```

```python
# Fill missing values and remove duplicate rows
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df = df.drop_duplicates()
print(df.isnull().sum())
```

```python
# Calculate summary statistics for Age and Fare
print("Mean:\n", df[["Age", "Fare"]].mean())
print("Median:\n", df[["Age", "Fare"]].median())
print("Standard Deviation:\n", df[["Age", "Fare"]].std())
print("Variance:\n", df[["Age", "Fare"]].var())
```

```python
# Plot a histogram of passenger ages
plt.hist(df["Age"], bins=10, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Histogram of Passenger Ages")
plt.show()
```

```python
# Plot a pie chart of passenger survival
df["Survived"].value_counts().sort_index().plot(
    kind="pie", labels=["Did not survive", "Survived"],
    autopct="%1.1f%%"
)
plt.title("Titanic Passenger Survival")
plt.ylabel("")
plt.show()
```
